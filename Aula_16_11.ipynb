```python
# Dicionários com Python
# 1 - Todo dicionário é um data frame (df)

turmaad6 = {
    # chave é unitário
    # valor por não ser unitário
    "professores" : ["Felipe","Luyz","Priscila"],
    "alunos": ["André","Paulo","Aline","Paola"]
}

print(turmaad6['professores'][1])
```

```python
# Criar as funções
# Ver os dados
# Adicionar novo professor
# Adicionar novo aluno
# Remover aluno
# Remover professor

def ver_dados():
  print('\n---- Professores cadastrados----')
  for i in turmaad6['professores']:
    print(i)

  print('\n---- Alunos cadastrados')
  for i in turmaad6['alunos']:
    print(i)

def add_prof():
  novo_professor = input('Informe o nome do professor: ')
  turmaad6['professores'].append(novo_professor)
  print(f'O professor {novo_professor} foi adicionado')
  ver_dados()

def add_aluno():
  novo_aluno = input('Informe o nome do aluno: ')
  turmaad6['alunos'].append(novo_aluno)
  print(f'O aluno {novo_aluno} foi adicionado')
  ver_dados()

def remover_aluno():
  posicao = 1
  for i in turmaad6['alunos']:
    print(f'{posicao} - {i}')
    posicao +=1
  remover = int(input('Informe a posição do aluno: '))
  # Para remover
  turmaad6["alunos"].pop(remover -1)
  ver_dados()

def remover_professor():
  posicao = 1
  for i in turmaad6['professores']:
    print(f'{posicao} - {i}')
    posicao +=1
  remover = int(input('Informe a posição do professor: '))
  turmaad6["professores"].pop(remover -1)
  ver_dados()
```
